```python
## IMPORT STATEMENTS

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
```

```python

Y = [10, 90]
V = [0,0]
dt = 0.01
nt = 100
f = []
var1 = []
var2 = []

for i in range(nt):
    f.append(Y.copy())
    a = -(Y[1]-Y[0])
    
    V[0] = V[0] - (a*dt)
    V[1] = V[1] + (a*dt)
    
    Y[0] = Y[0]+V[0]
    Y[1] = Y[1]+V[1]

    var1.append(Y[0])
    var2.append(Y[1])


    
```

```python
fig, ax = plt.subplots()


bars = ax.bar(["C[0]", "C[1]"], f[0])

ax.set_ylim(0, 100)
ax.set_ylabel("Value")
ax.set_title("Values of C")

# Animation function
def update(frame):
    bars[0].set_height(f[frame][0])
    bars[1].set_height(f[frame][1])
    ax.set_title(f"Values of C — Step {frame}")

    return bars

ani = FuncAnimation(
    fig,
    update,
    frames=nt,
    interval=100,
    blit=True,
    repeat=True
)

ani.save("2CWM.gif", writer="pillow", fps=30)

plt.show()
```
